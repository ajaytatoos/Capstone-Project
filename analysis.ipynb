```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("books.db")

print("Database connected successfully!")
```

```python
books_df = pd.read_sql("""
SELECT
    books.book_id,
    books.title,
    books.price_gbp,
    books.rating,
    books.in_stock,
    books.price_inr,
    categories.category_name
FROM books
JOIN categories
    ON books.category_id = categories.category_id
""", conn)

books_df.head()
```

```python
books_by_category = (
    books_df.groupby("category_name")
    .size()
    .sort_values(ascending=False)
)

print("Pandas Query 1 - Books by category:")
print(books_by_category)
```

```python
average_price_by_category = (
    books_df.groupby("category_name")["price_gbp"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Pandas Query 2 - Average price by category:")
print(average_price_by_category)
```

```python
top_5_expensive = (
    books_df.sort_values("price_gbp", ascending=False)
    [["title", "price_gbp", "rating", "category_name"]]
    .head(5)
)

print("Pandas Query 3 - Top 5 most expensive books:")
top_5_expensive
```

```python
average_rating_by_category = (
    books_df.groupby("category_name")["rating"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Pandas Query 4 - Average rating by category:")
print(average_rating_by_category)
```

```python
in_stock_by_category = (
    books_df[books_df["in_stock"] == 1]
    .groupby("category_name")
    .size()
    .sort_values(ascending=False)
)

print("Pandas Query 5 - Books currently in stock by category:")
print(in_stock_by_category)
```

```python
books_table = pd.read_sql("SELECT * FROM books", conn)

categories_table = pd.read_sql("SELECT * FROM categories", conn)

print("Books table shape:", books_table.shape)
print("Categories table shape:", categories_table.shape)
```

```python
merged_df = books_table.merge(
    categories_table,
    on="category_id",
    how="inner"
)

print("Merged DataFrame shape:", merged_df.shape)

merged_df.head()
```
